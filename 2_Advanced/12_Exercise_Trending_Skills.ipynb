```python
# Importing Libraries
import ast
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_data_analyst = df[df['job_title_short'] == 'Data Analyst'].copy()
df_data_analyst
```

```python
df_data_analyst['job_posted_month_no'] = df_data_analyst['job_posted_date'].dt.month
df_data_analyst
```

```python
df_data_analyst_exploded = df_data_analyst.explode('job_skills')
df_data_analyst_exploded
```

```python
df_data_analyst_exploded.pivot_table(index = 'job_posted_month_no', columns = 'job_skills', aggfunc = 'size')
```

```python
# Fill the 'NaN' values with 0 so there's no discrepancy while plotting the graph

df_data_analyst_pivot = df_data_analyst_exploded.pivot_table(index = 'job_posted_month_no', columns = 'job_skills', aggfunc = 'size', fill_value = 0)
df_data_analyst_pivot.plot(kind = 'line')
df_data_analyst_pivot
```

```python
# Need to create a row that sums up the values of every skill for the entire year
df_data_analyst_pivot.loc['Total'] = df_data_analyst_pivot.sum()
df_data_analyst_pivot
```

```python
df_data_analyst_pivot.loc['Total'].sort_values(ascending = False)
```

```python
df_data_analyst_pivot = df_data_analyst_pivot[df_data_analyst_pivot.loc['Total'].sort_values(ascending = False).index]
```

```python
df_data_analyst_pivot = df_data_analyst_pivot.drop('Total')
df_data_analyst_pivot
```

```python
df_data_analyst_pivot.iloc[:, :5].plot(kind = 'line')
plt.title('Top 5 Skills for Data Analysts per month')
plt.xlabel('')
plt.ylabel('Count')
plt.show()
```
