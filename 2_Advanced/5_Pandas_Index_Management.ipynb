```python
# Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
```

## reset_index()

```python
df_india = df[df['job_country'] == 'India']
df_india.head()
```

```python
# inplace: Whether to modify the dataframe rather than creating a new one

df_india.reset_index(inplace = True)

df_india.head()
```

## set_index()

Set the dataframe index using the existing columns

```python
df_india.set_index('job_index', inplace = True)

df_india.head()
```

## sort_index()

Returns a new dataframe sorted by label if 'inplace' argument is 'False', otherwise updates the original dataframe and returns None

```python
median_pivot = df_india.pivot_table(index = 'job_title_short', values = 'salary_year_avg', aggfunc = ['median', 'min', 'max'])
median_pivot
```

```python
median_pivot.sort_values(by = ('median', 'salary_year_avg'), ascending = False)
```

```python
# Now using the sort_index() function

median_pivot.sort_index()
```
