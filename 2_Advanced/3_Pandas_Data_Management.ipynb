```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
```

```python
# To return the same result everytime you use the sample() method, specifiy the 'random_state' parameter

df.sample(10, random_state = 32)
```

```python
# To make the a of an object's indices and data, use the copy() method

dataset = load_dataset('lukebarousse/data_jobs')
df_original = dataset['train'].to_pandas()
```

```python
df_original
```

```python
df_altered = df_original

df_altered.loc[:5, 'salary_year_avg']
```

```python
median_salary = df_altered['salary_year_avg'].median()

median_salary
```

```python
df_altered['salary_year_avg'] = df_altered['salary_year_avg'].fillna(median_salary)
```

```python
df_altered.loc[:5, 'salary_year_avg']
```

```python
df_original.loc[:5, 'salary_year_avg']
```

```python
print(id(df_original))
print(id(df_altered))
```

Hence from above, we conclude that the operations performed on df_altered were being performed over df_original as well which is a bad practice
