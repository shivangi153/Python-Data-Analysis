## Data Cleaning

```python
# job_posted_date is an object right now; convert it to datetime datatype

print(df.job_posted_date[0])
print(type(df.job_posted_date[0]))
```

```python
pd.to_datetime(df.job_posted_date)

# Assign it to itself

df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
```

```python
# but the job_posted_date is still an object

df.info()
```

```python
# Hence, re-assign the job_posted_date column to itself after it has been converted to the corrected datatype

# Now, after conversion to datetime, different datetime methods can now be used over this column

df.job_posted_date.dt.month
```

```python
# Now, create a new column named job_posted_month to extract the values fetched above

df['job_posted_month'] = df.job_posted_date.dt.month
```

```python
# To sort the jobs on the basis of the job_posted_date

df.sort_values('job_posted_date')
```

```python
# To sort the values and assign them back to the original dataframe:
# Method-1:
# df = df.sort_values('job_posted_date')

# OR

# Method-2:
df.sort_values(by = 'job_posted_date', inplace = True)

df
```

```python
# Further cleaning:
# (i)  Remove the salary_hour_date column
# (ii) Remove the rows from the job_posted_date column that do not have any values in them or are NaN

# (i)
# axis = 1 for columns; axis = 0 for index; by default, axis = 0
df.drop(labels='salary_hour_avg', axis=1, inplace = True)
```
