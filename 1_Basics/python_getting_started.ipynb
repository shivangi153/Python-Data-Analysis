## == and is Operators

The core difference is that == compares values, while is compares identities

- == (Equality Operator): Checks if the data or contents of two objects are the same.
- is (Identity Operator): Checks if two variables point to the exact same memory location

```python
# Example
luke_salary = 100_000
kelly_salary = 100_000

print(luke_salary == kelly_salary)
print(luke_salary is kelly_salary)
```

```python
print(id(luke_salary))
print(id(kelly_salary))
```

```python
# Write a Python script to find out what jobs you are qualified based on a set of skills.

my_skills = ['Python', 'SQL', 'Excel']

job_roles = [
    {'role': 'Data Analyst', 'skills': ['Python', 'SQL', 'Excel']},
    {'role': 'Data Scientist', 'skills': ['Python', 'R', 'Machine Learning', 'Deep Learning']},
    {'role': 'Machine Learning Engineer', 'skills': ['Python', 'TensorFlow', 'PyTorch', 'Scikit-Learn']},
    {'role': 'Data Engineer', 'skills': ['Python', 'Apache Spark', 'Hadoop', 'SQL']},
    {'role': 'Business Intelligence Analyst', 'skills': ['Python', 'SQL', 'Tableau', 'Power BI', 'Excel']},
    {'role': 'Quantitative Analyst', 'skills': ['R', 'Python', 'MATLAB', 'Statistics']},
    {'role': 'Operations Analyst', 'skills': ['Python', 'SQL', 'Data Visualization', 'Process Improvement']},
    {'role': 'Database Administrator', 'skills': ['SQL', 'Oracle', 'MySQL', 'Database Management']},
    {'role': 'AI Engineer', 'skills': ['Python', 'TensorFlow', 'PyTorch', 'Computer Vision']},
    {'role': 'Statistician', 'skills': ['R', 'SAS', 'Python', 'Statistical Modeling']}
]
```

```python
# Method-1:

for job in job_roles:
  if ('Python' in job['skills']) and ('SQL' in job['skills']) and ('Excel' in job['skills']):
    print(job['role'])
```

```python
# Method-2:

for job in job_roles:
  required_skills = job['skills']

  if (set(my_skills).issubset(required_skills)):
    print(job['role'])
```

```python
# Method-3:

for job in job_roles:
  qualified = True

  for skill in my_skills:
    if(skill not in job['skills']):
      qualified = False

  if qualified:
    print(job['role'])
```

```python
# Method-3:

qualified_roles = []

for job in job_roles:
  qualified = True

  for skill in my_skills:
    if(skill not in job['skills']):
      qualified = False

  if qualified:
    qualified_roles.append(job['role'])

qualified_roles
```

# Lambda Functions

A small, anonymous function that is defined without a name. Unlike standard functions created using the 'def' keyword, lambda functions are written in a single line and are typically used for quick, one-time operations.

```python
mul_two = lambda x: x * 2
mul_two(3)
```

```python
(lambda x: x * 2)(3)
```

```python
(lambda x, y: x * 2 + y)(3, 4)
```

```python
# Calculating the bonus-added salaries using list comprehension

salary_list = [100000, 200000, 150000, 120000, 80000, 750000]

def calculate_salary(base, bonus = 0.1):
  return base * (1 + bonus)

bonus_salary = [calculate_salary(salary) for salary in salary_list]
bonus_salary
```

```python
# Defining the above process using lambda function

salary_list = [100000, 200000, 150000, 120000, 80000, 750000]

def calculate_salary(base, bonus = 0.1):
  return base * (1 + bonus)

bonus_salary = [(lambda x: x * 1.1)(salary) for salary in salary_list]
bonus_salary
```
