```python
import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df.info()
```

```python
df.count()
```

```python
df['job_title_short'].unique()
```

```python
df['job_title_short'].value_counts()
```

```python
# To find the minimum salary of all the job_title_short values

df.groupby('job_title_short')['salary_year_avg'].min()
```

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

```python
df.groupby(['job_title_short', 'job_country'])['salary_year_avg'].min()
```

```python
df.groupby('job_title_short')[['salary_year_avg', 'salary_hour_avg']].median()
```

```python
df.groupby('job_title_short')['salary_year_avg'].agg(['min', 'max', 'mean', 'median'])
```
